```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import plotly as plty
import seaborn as sns
df = pd.read_csv("1314-003_police_officers_rank_gender_csv_2008-12.csv", encoding = "ISO-8859-1")
df_long = df.melt(
    id_vars=['Geography','Rank','Sex'],             # Columns to keep as they are
    value_vars=['2008', '2009', '2010', '2011', '2012'], 
    var_name='Year',                 # The name for the new column of headers
    value_name='Number'              # The name for the new column of data values
)
df_long['Year'] = df_long['Year'].astype(int)
df_long = df_long.sort_values(by='Year')
df_long = df_long.dropna(subset=['Number'])
df_provinces = df_long[df_long['Geography'] != 'Canada']
# df_long = df_long[df_long['Rank']=='Total number of police officers (1)']
# df_canada = df_long[df_long['Geography'] == 'Canada']
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Set up the figure size and style
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_facecolor('white')  # Apply your grey background

# 2. Plot the provincial totals 
# estimator='sum' forces Seaborn to add the ranks together for each year/sex/province
sns.lineplot(
    data=df_provinces,
    x='Year',
    y='Number',
    hue='Geography',       # Distinct colors for each province
    style='Sex',           # Distinct line styles (solid vs dashed) for Male/Female
    markers=True,          # Adds the dots to each year
    errorbar=None,         # Removes the shadow/confidence interval band
    ax=ax
)

# 3. Clean up the X-Ticks to be ONLY your target years
ax.set_xticks([2008, 2009, 2010, 2011, 2012])
ax.set_xticklabels([2008, 2009, 2010, 2011, 2012])

# 4. Format labels and title
ax.set_title("Total Personnel Trends by Province and Sex (2008-2012)", fontsize=12, weight='bold')
ax.set_xlabel("Year")
ax.set_ylabel("Number of Personel")

# 5. Position the legend neatly outside the chart to the right
ax.legend(
    loc='upper left', 
    bbox_to_anchor=(1, 1), 
    fontsize=9,   
    handletextpad=0.5
)

plt.tight_layout()
plt.show()
```

I want to make this interactive since it will be easier to understand and see all of the data. The current version cannot show the individual regions and they all get pushed together at the bottom. 

```python
totals = df_long[df_long["Rank"] == "Total number of police officers (1)"] # separates the males and females
males   = totals[totals["Sex"] == "Males"].copy()
females = totals[totals["Sex"] == "Females"].copy()

geographies = df["Geography"].unique().tolist() # make sure that they are all separated
default_geo = "Canada"

m0 = males[males["Geography"] == default_geo].sort_values("Year") # sorts them by year so that they will show up correctly 
f0 = females[females["Geography"] == default_geo].sort_values("Year")
```

```python
import plotly.graph_objects as go

fig = go.Figure() # creates a base figure to work from

fig.add_trace(go.Scatter( # traces allow the male and female groups to be separated and then individually formatted
    x=m0["Year"], y=m0["Number"],
    name="Males",
    mode="lines+markers",
    line=dict(color="#0072B2", width=2.5),
    marker=dict(size=8),
))

fig.add_trace(go.Scatter(
    x=f0["Year"], y=f0["Number"],
    name="Females",
    mode="lines+markers",
    line=dict(color="#D55E00", width=2.5),
    marker=dict(size=8),
))
```

```python
buttons = [] # this will allow the changing of geography and the updating 
for geo in geographies:
    m = males[males["Geography"] == geo].sort_values("Year") # want to sort them to ensure that the show up in order
    f = females[females["Geography"] == geo].sort_values("Year")
    buttons.append(dict(
        label=geo,
        method="update", 
        args=[
            {
                "x": [m["Year"].tolist(), f["Year"].tolist()],
                "y": [m["Number"].tolist(), f["Number"].tolist()],
            },
            {"title": {"text": f"Police Personnel by Sex — {geo}", "x": 0.5}}, 
        ],
    )) # this should update the title after the selection is changed
```

```python
# layout of the figure

fig.update_layout( 
    title=dict(text=f"Police Personnel by Sex — {default_geo}", x=0.5), # adds titles in for information
    xaxis=dict(title="Year", tickvals=[2008, 2009, 2010, 2011, 2012], dtick=1), # ensures there is only 1 tick per year
    yaxis=dict(title="Number of Personnel"),
    legend=dict(title="Sex"),
    updatemenus=[dict(
        buttons=buttons,
        direction="down",
        showactive=True,
        x=0.10, xanchor="left", # anchors the drop down button
        y=1.19, yanchor="top",
        bgcolor="white",
        bordercolor="#cccccc")],
    annotations=[
        dict(
            text="Geography:", 
            x=-0.02, y=1.17, # anchors the geography label for the drop
            xref="paper", yref="paper",
            showarrow=False, align="left",
            font=dict(size=14)
            ),
        dict(
            text="Source: Ministry of Solicitor Services Information via ontario.ca",
            x=0.7, y=-.3,
            xref="paper", yref="paper",
            showarrow=False, xanchor="left", align="right",
            font=dict(size=12)
            )
    ],
    template="plotly_white",
    margin=dict(t=120),
    
)
```

```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.ticker as mticker
df = pd.read_csv("Fire_Data.csv")
df
```

```python
df_clean = df.melt(
    id_vars=['Cause'],             # Columns to keep as they are
    value_vars=['1990','1991','1992','1993','1994','1995','1996','1997','1998','1999',
                '2000','2001','2002','2003','2004','2005','2006','2007','2008','2009',
                '2010','2011','2012','2013','2014','2015','2016','2017','2018', '2019',
                '2020','2021','2022','2023','2024'], 
    var_name='Year',                 # The name for the new column of headers
    value_name='Number of Fires'              # The name for the new column of data values
)
```

```python
df_clean
```

```python
causes_long = (
    df[df["Cause"] != "Annual Total"]
    .melt(id_vars="Cause", var_name="Year", value_name="Count")
)
causes_long["Year"]  = causes_long["Year"].astype(int)
causes_long["Count"] = causes_long["Count"].str.replace(",", "").astype(int)
```

```python

causes_pivot = causes_long.pivot(index="Year", columns="Cause", values="Count")
causes_pivot = causes_pivot[["Human activity", "Natural cause", "Unspecified"]]


causes_pct = causes_pivot.div(causes_pivot.sum(axis=1), axis=0) * 100
print(causes_pct.head())
```

```python
sns.set_theme(style="whitegrid")  # 
 
palette = { # colour blind friendly
    "Human activity": "#0072B2",  # Okabe-Ito blue
    "Natural cause":  "#D55E00",  # Okabe-Ito vermillion
    "Unspecified":    "#999999",  # grey
}
```

```python
# Melt the percentage pivot back to long format
causes_pct_long = causes_pct.reset_index().melt(
    id_vars="Year",
    var_name="Cause",
    value_name="Percentage"
)

# Plot
fig, ax = plt.subplots(figsize=(16, 5)) 

sns.barplot(
    data=causes_pct_long,
    x="Year", y="Percentage", hue="Cause",
    palette=palette,
    ax=ax
)

ax.set_ylim(0, 80)
ax.set_ylabel("Share of Fires (%)", fontsize=12)
ax.set_xlabel("Year", fontsize=12)
ax.set_title("Canadian Wildfires by Cause (% of Annual Total), 1990–2024", fontsize=13)
ax.tick_params(axis="x", rotation=45)
ax.legend(title="Cause", framealpha=0.9)

plt.tight_layout()
```

This is too crowded to see. So I will stack the bars and add in the number of fires that year. 

```python
# ── 1. Load and reshape ───────────────────────────────────────────────────────
df = pd.read_csv("Fire_Data.csv")

causes_long = ( # melt these as they contain the total as well
    df[df["Cause"] != "Annual Total"] # remove the total that is there 
    .melt(id_vars="Cause", var_name="Year", value_name="Count") # 
)
causes_long["Count"] = causes_long["Count"].str.replace(",", "").astype(int) # these were recorded as strings so the commas were removed and the str are converted to int 

total_long = ( # a separate one for total 
    df[df["Cause"] == "Annual Total"]
    .melt(id_vars="Cause", var_name="Year", value_name="Count")
)
total_long["Year"]  = total_long["Year"].astype(int)
total_long["Count"] = total_long["Count"].str.replace(",", "").astype(int) # same process as above 

# ── 2. Calculate percentages ──────────────────────────────────────────────────
causes_pivot = causes_long.pivot(index="Year", columns="Cause", values="Count")
causes_pivot = causes_pivot[["Human activity", "Natural cause", "Unspecified"]] 

causes_pct = causes_pivot.div(causes_pivot.sum(axis=1), axis=0) * 100 # calculate the percentages since these are better to use for main message

# ── 3. Plot ───────────────────────────────────────────────────────────────────
sns.set_theme(style="whitegrid")  
palette = {
    "Human activity": "#0072B2",  # Okabe-Ito blue
    "Natural cause":  "#D55E00",  # Okabe-Ito vermillion
    "Unspecified":    "#999999",  # grey
}

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 10), sharex=False) # creates the plate with 2 graphs

# ── Top: stacked bar chart (% per year) ──────────────────────────────────────
causes_pct.plot(
    kind="bar", stacked=True,
    color=[palette["Human activity"], palette["Natural cause"], palette["Unspecified"]],
    ax=ax1, alpha=0.85, width=0.8
)
ax1.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x:.0f}%")) # maintains the y axies
ax1.set_ylim(0, 100)
ax1.set_ylabel("Share of Fires (%)", fontsize=12)
ax1.set_title("Canadian Wildfires by Cause, 1990–2024", fontsize=14, pad=10) # title
ax1.legend(loc="upper right", framealpha=0.9, fontsize=10)
ax1.set_xticks(range(0, len(causes_pct), 5))
ax1.set_xticklabels(causes_pct.index[::5], rotation=45)
ax1.set_xlabel("")
ax1.grid(axis="x", alpha=0.3)

sns.lineplot( # this lineplot is to show the total number of fires that year (for the data-savvy, helps to show the actual number of fires)
    data=total_long, x="Year", y="Count", ax=ax2, 
    color="black", linewidth=2, marker="o", markersize=4,
) 
ax2.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{int(x):,}")) # sets the y-axies
ax2.set_ylabel("Annual Total Fires", fontsize=12)  # sets the labels 
ax2.set_xlabel("Year", fontsize=12)
ax2.set_title("Annual Total Wildfires", fontsize=12, pad=6)
ax2.grid(axis="x", alpha=0.3)

# ── Source note ───────────────────────────────────────────────────────────────
fig.text(0, -0.02, "Source: Canadian National Fire Database via ontario.ca", # the source so that it is more trustworthy, replicable 
         fontsize=9, color="#666666")

plt.tight_layout()
```
